```python
import pandas as pd
```

```python
# Create a small employee performance dataset
employees_df = pd.DataFrame({
    'employee': ['Anna', 'Ben', 'Clara', 'Daniel', 'Eva', 'Frank', 'Grace', 'Henry'],
    'department': ['Sales', 'IT', 'Sales', 'HR', 'IT', 'HR', 'Sales', 'IT'],
    'hours_worked': [38, 40, 36, 35, 42, 37, 39, 41],
    'tasks_completed': [28, 35, 31, 24, 38, 26, 33, 36],
    'performance_score': [82, 91, 88, 79, 94, 81, 90, 89]
})

employees_df.head()
```

```python
# Group employees by department
department_groups = employees_df.groupby('department')

# Calculate the mean of numeric columns for each department
department_groups.mean(numeric_only=True)
```
