# Removing Duplicate Data in Pandas

Duplicate records can affect data quality and lead to incorrect analysis.
In this example, I practise identifying and removing duplicate customer records.

```python
customers_df = pd.DataFrame({
    'customer_id': [101, 101, 102, 102, 103, 103, 103],
    'name': ['Emma', 'Emma', 'Noah', 'Noah', 'Olivia', 'Olivia', 'Olivia'],
    'city': ['London', 'London', 'Manchester', 'Manchester',
             'Birmingham', 'Birmingham', 'Birmingham']
})

customers_df
```

### Identifying duplicate rows

`duplicated()` checks whether a row has already appeared in the DataFrame.
It returns `True` for duplicate rows and `False` for unique rows.

```python
# Check which rows are duplicates
customers_df.duplicated()
```

```python
# Count the total number of duplicate rows
customers_df.duplicated().sum()
```

### Removing exact duplicates

`drop_duplicates()` removes rows where all values are duplicated.
By default, Pandas keeps the first occurrence.

```python
customers_df.drop_duplicates()
```

```python
clean_customers_df = customers_df.drop_duplicates()

clean_customers_df
```

```python
customer_records = pd.DataFrame({
    'customer_id': [101, 101, 102, 102, 103, 103, 104],
    'name': ['Emma', 'Emma', 'Noah', 'Noah', 'Olivia', 'Olivia', 'Liam'],
    'city': ['London', 'London', 'Manchester', 'Leeds',
             'Birmingham', 'Birmingham', 'Oxford']
})

customer_records
```

```python
customer_records.drop_duplicates()
```

```python
# Remove duplicates based only on customer_id
unique_customers = customer_records.drop_duplicates(subset=['customer_id'])

unique_customers
```
