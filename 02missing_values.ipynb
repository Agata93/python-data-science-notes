## Working with Missing Data in Pandas

```python
import numpy as np
import pandas as pd

from pandas import DataFrame
```

### Filling missing values using fillna(), replace() and interpolate()

```python
data = {
    'name': ['Anna', 'Tom', 'Emily', 'David', 'Laura', 'James', 'Sophie'],
    'age': [25, 31, 28, 35, 29, 33, 27],
    'department': ['Sales', 'IT', 'HR', 'Finance', 'IT', 'Sales', 'HR'],
    'score': [78, 91, 85, 74, 88, 80, 95]
}

employees_df = DataFrame(data)

# Add missing values to some cells for practice
employees_df.iloc[2:5, 1] = np.nan
employees_df.iloc[3:6, 3] = np.nan
employees_df.iloc[3, :] = np.nan

employees_df
```

```python
# Check which values are missing
employees_df.isnull()
```

```python
# Check which values are not missing
employees_df.notnull()
```

```python
# Create a boolean Series showing where values in the age column are missing
bool_series = pd.isnull(employees_df['age'])

# Display only rows where age is missing
employees_df[bool_series]
```

```python
# Replace all missing values with 0
employees_df.fillna(0)
```

Fill missing values using the previous valid observation

```python
employees_df.ffill()
```

Fill missing values using the next valid observation

```python
employees_df.bfill()
```

Fill missing values in 'rank' using linear interpolation

```python
employees_df['score'] = employees_df['score'].interpolate(method='linear')

employees_df
```

```python
# Remove rows containing at least one missing value
employees_df.dropna()
```

```python
# Remove rows only if all values in the row are missing
employees_df.dropna(how='all')
```

```python
# Remove columns containing missing values
employees_df.dropna(axis=1)
```

```python
# Remove rows containing missing values
employees_df.dropna(axis=0)
```
