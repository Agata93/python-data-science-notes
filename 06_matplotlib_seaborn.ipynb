```python
import matplotlib.pyplot as plt
import seaborn as sns
from pandas import DataFrame
```

```python
# Create a dictionary containing fictional employee sales data
data = {
    'employee': ['Anna', 'Daniel', 'Olivia', 'James', 'Sophie', 'Adam', 'Emily'],
    'monthly_sales': [12500, 9800, 14200, 8700, 13100, 10500, 11800],
    'department': [
        'Electronics', 'Clothing', 'Electronics',
        'Home', 'Clothing', 'Home', 'Electronics'
    ],
    'performance_rank': [3, 6, 1, 7, 2, 5, 4]
}
# Convert the dictionary into a pandas DataFrame
df = DataFrame(data)

# Display the DataFrame
df
```

```python
# Set the figure size
plt.figure(figsize=(8, 5))

# Create a bar chart showing monthly sales for each employee
plt.bar(df['employee'], df['monthly_sales'], color='steelblue')

# Add the axis labels and chart title
plt.xlabel('Employee')
plt.ylabel('Monthly Sales (£)')
plt.title('Monthly Sales by Employee')
# Rotate employee names to improve readability
plt.xticks(rotation=45)
#Adjust the layout and display the chart
plt.tight_layout()
plt.show()
```

```python
# Create a Seaborn bar chart grouped by department
plt.figure(figsize=(8, 5))

plot = sns.barplot(
    data=df,
    x='employee',
    y='monthly_sales',
    hue='department',
    palette='pastel'
)

# Add the title and axis labels
plot.set_title('Monthly Sales by Employee')
plot.set_xlabel('Employee')
plot.set_ylabel('Monthly Sales (£)')

# Improve the layout and display the chart
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Sort employees by performance rank
df_sorted = df.sort_values('performance_rank')

# Create a line plot of monthly sales
plt.figure(figsize=(8, 5))

plt.plot(
    df_sorted['employee'],
    df_sorted['monthly_sales'],
    marker='o',
    color='darkgreen'
)

# Add the title, axis labels and grid
plt.xlabel('Employee')
plt.ylabel('Monthly Sales (£)')
plt.title('Monthly Sales by Performance Rank')

# Improve the layout and display the chart
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    df['monthly_sales'],
    labels=df['employee'],
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Share of Total Sales by Employee')
plt.show()
```

```python
# Create a colour palette with one colour for each employee
colors = sns.color_palette('pastel', n_colors=len(df))

# Create a pie chart showing each employee's share of total sales
plt.figure(figsize=(7, 7))

plt.pie(
    df['monthly_sales'],
    labels=df['employee'],
    colors=colors,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Share of Total Sales by Employee')
plt.show()
```
