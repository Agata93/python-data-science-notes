## Data Filtering and Selection

```python
import numpy as np
import pandas as pd

from pandas import DataFrame
```

```python
numbers_df = DataFrame(
    np.arange(10, 70, 2).reshape(10, 3),
    index=['sample 1', 'sample 2', 'sample 3', 'sample 4', 'sample 5',
           'sample 6', 'sample 7', 'sample 8', 'sample 9', 'sample 10'],
    columns=['feature A', 'feature B', 'feature C']
)

numbers_df
```

```python
numbers_df
```

```python
numbers_df.iloc[0,1]
```

#### Comparison operators (> < = <= => == !=) and Masking.

Create a boolean mask to identify values greater than 42

```python
mask = numbers_df > 42
mask
```

```python
numbers_df[mask]
```

```python
numbers_df[numbers_df>42]=0
numbers_df
```
